
# DELTA File Format

```python
from pyspark.sql import functions as F
```

```python
df = spark.table("datasets.apple_global_sales_dataset")

df.display()
```

```python
df.write.format("delta").mode("overwrite").saveAsTable("datasets.apple_sales_delta")
```

```python
spark.sql("DESCRIBE DETAIL datasets.apple_sales_delta").select("format", "location", "numFiles", "properties").display()
```

```python
df = spark.table("datasets.apple_sales_delta")
df.display()
```

```python
spark.sql("DESCRIBE HISTORY datasets.apple_sales_delta").display()
```

# Optimizations in delta Table
<p>1. OPtimize</p>
<p>2. Z-Order</p>

```python
%sql
-- OPTIMIZE reorganizes the data files for better performance and reduced latency.
-- ZORDER BY (year, month) colocates data by year and month to improve query speed on these columns.
OPTIMIZE datasets.apple_sales_delta
ZORDER BY (year, month)
```
