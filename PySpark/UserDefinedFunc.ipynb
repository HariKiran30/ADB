
# _____USER DEFINED FUNCTIONS____________

<p>Custom Functions that allows you to extent spark in built capabilities by applying own python logic</p>
<p>It is slow for Large datasets</p>
<p>Not commonly used </p>


```python
from pyspark.sql import functions as F
from pyspark.sql import types as T
```

```python
cus_df = spark.table("samples.tpch.customer")
cus_df.display()
```

```python
#Method 1
def make_it_to_lower(name:str):
    return name.lower()

make_it_to_lower_udf = F.udf(make_it_to_lower, T.StringType())
```

```python
df = cus_df.withColumn("c_mktsegment", make_it_to_lower_udf(F.col("c_mktsegment")))
df.display()
```

```python
#Method 2
@udf(T.StringType())
def make_it_lower2(name: str):
    return name.lower()
```

```python
df1 = cus_df.withColumn("c_mktsegment", make_it_lower2(F.col("c_mktsegment")))
df1.display()
```
