```python
trip_df = spark.table('samples.nyctaxi.trips')
trip_df.display()
```

```python
# with column functions
from pyspark.sql import functions as F
trip_df = trip_df.withColumn('fare_amount_INR',F.col('fare_amount')*92)
trip_df.display()
```

```python
# 

from pyspark.sql import functions as F
trip_df = trip_df \
    .withColumn('fare_amount_INR', F.col('fare_amount') * 92) \
    .withColumn('dummy1', F.lit('Dummy'))\
     .withColumn('dummy2', F.lit('Dummy'))


trip_df.display()
```

```python
trip_df = trip_df.withColumnRenamed("fare_amount_INR","fare_amount_Rupee")
trip_df.display()
```

```python
trip_df = spark.sql("SELECT * FROM samples.nyctaxi.trips")
trip_df.display()
```

```python
#Method 1 of selecting Columns
df = trip_df.select("tpep_pickup_datetime","tpep_dropoff_datetime")
df.show()
```

```python
#Method 2
df = trip_df.select(trip_df.tpep_pickup_datetime,trip_df.tpep_dropoff_datetime)
df.show()
```

```python
#Method 3
df = trip_df.select(trip_df["tpep_pickup_datetime"], trip_df["tpep_dropoff_datetime"])

df.show()
```

```python
# Method 4 (Standard)
from pyspark.sql import functions as F

df = trip_df.select(F.col("tpep_pickup_datetime"),F.col('tpep_dropoff_datetime'))

df.show()
```
