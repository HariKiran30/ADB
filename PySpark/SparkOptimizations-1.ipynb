# Spark Join Strategies
<p>Sort merge Join - Defualt</p>
<p>Shuffle Hash Join</p>
<p>BroadCast Join - Most Optimised Join In Spark</p>


```python
cus_df = spark.table("samples.tpch.customer")
cus_df.display()
```

```python
 orders_df= spark.table("samples.tpch.orders")
 orders_df.display()
```

```python
# Join Order df and customer df to get order Information

# Broadcast Join
from pyspark.sql import functions as F

boradcast_df = (
    orders_df.
    join(
        F.broadcast(cus_df), orders_df.o_custkey == cus_df.c_custkey,
                   how = 'left'
                   )
)

boradcast_df.display()
```

```python
boradcast_df.count()
```

```python
# Default Sort Merge Join
sort_merge_df = (
    orders_df.join(
        cus_df,
        orders_df.o_custkey == cus_df.c_custkey,
        how='left'
    )
)

sort_merge_df.explain()
display(sort_merge_df)
```

```python
# Shuffle Hash Join
shuffle_hash_df = (
    orders_df.join(
        cus_df.hint("shuffle_hash"),
        orders_df.o_custkey == cus_df.c_custkey,
        how='left'
    )
)

shuffle_hash_df.explain()
display(shuffle_hash_df)
```
